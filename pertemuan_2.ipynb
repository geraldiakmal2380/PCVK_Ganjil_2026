# 4. Menampilkan Citra greyscale

```python
img_grey = cv2.imread("/content/drive/MyDrive/Praktikum PCVK 2026/KTM", cv2.IMREAD_GRAYSCALE)

#tanpa cmap (cmap adalah library dari matplotlib fungsinya ngatur uhhhh colormaps???? gtw maksudnya apa but ok)
plt.subplot(1,2,1) #ini untuk apa ya (kata dokumentasinya buat ngatur ncolnrows=1, ncols=1 smth.)
plt.imshow(img_grey)
plt.title("Tanpa cmap")


#dengan cmap (yay gak ke deep fried)
plt.subplot(1,2,2) #pokoknya angka kanan harus 2. Kyk nya malah timpang tindih deh kalo angkanya 1
plt.imshow(img_grey, cmap="gray")#cmap nya bisa juga diganti dari gray ke magma biar ke deep fried
plt.title("dengan cmap")
```

# 5. Melakukan Resizing
