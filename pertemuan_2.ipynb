# Membuat Geometri 2D dengan openCV

```python
black_img = np.zeros(shape=(530,530,3),dtype=np.int16)
#3 itu buat menspesifikasikan channel warna. Jika 3 hilang ya gambar hanya akan hitam putih (Grayscale)
#datatype int16 agar bisa angka bisa negatif. soalnya by default numpy pakenya itu Unsigned integer 8 bit. Artinya angka hanya bisa dari 0 - 255 saja (Yes tidak bisa negatif).
plt.imshow(black_img)
```

Berikutnya tambahkan kotak berwarna biru

```python
cv2.rectangle(black_img, pt1=(290,0), pt2=(520,130), color=(0,0,255), thickness=10)#pt1 & pt2 itu kordinat kiri atas dan kanan bawah. Thickness itu ketebalan garis (GIMP is free bro)
plt.imshow(black_img)
```

Buat Persegi merah (yummy)

```python
cv2.rectangle(black_img, pt1=(250,250), pt2=(350,350), color=(255,0,0), thickness=15)
plt.imshow(black_img)
```

Buat Lingkaran berwarna hijau (cuz why not)

```python
cv2.circle(black_img, center=(150,150), radius=60, color=(0,250,0), thickness=8)
plt.imshow(black_img)
```

Tambahkan garis miring dari kiri atas ke kanan bawah dan lingkaran di area sekitar kanan bawah

```python
cv2.circle(black_img, center=(400,400), radius=50, color=(0,255,0), thickness=-1)#Ini buat lingkaran
cv2.line(black_img, pt1=(0,0), pt2=(530,530), color=(255,255,255), thickness=5)#ini buat garis
plt.imshow(black_img)
```

Tambahkan tulisan ke dalam gambar

```python
font=cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(black_img, text="PCVK is fun!", org=(10,500), fontFace=font, fontScale=2, color=(255,255,0), thickness=3, lineType=cv2.LINE_AA)
#lineType = line_AA berfungsi untuk anti aliasing tulisan
#fontFace buat menentukan gaya font nya (cek variabel font di atas)
plt.imshow(black_img)
```
