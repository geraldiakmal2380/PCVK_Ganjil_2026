```python
#Fungsi untuk show gambar
def show(img, title=''):
    plt.figure(figsize=(8, 6))
    if img.ndim == 2:
        plt.imshow(img, cmap='gray')
    else:
        plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
    plt.title(title); plt.axis('off'); plt.show()
```
