Buat Fungsi convolution

```python
def convolution2d(image, kernel, stride, padding):
  '''
  Apa fungsi fungsi ini?
  ----------------------
  Fungsi ini berfungsi untuk melakukan convolution pada citra.
  convolution atau konvulsi adalah proses pengkalian antara elemen demi elemen antar kernel dan region yang bersesuaian. Kemudian dijumlahkan

  Apa fungsi paraeter di atas?
  ----------------------------
  image   : ini image yang kita buat di awal tadi
  kernel  : Filter yang akan di geser ke seluruh gambar (Jujur agak susah memahaminya)
  Padding : Jarak antara kernel dan gambar
  Stride  : Jarak yang akan ditempuh kernel (biasanya 1).


  Gimana cara alur prosesnya ?
  ----------------------------
  1. Tambahkan padding jika diperlukan
  2. Hitung ukuran output ((H + 2*padding - K) /stride) + 1
  3. Geser kernel dari kiri atas ke kanan bawah
  4. Di setiap posisi dan pergeseran, lakukan :
      - Ambil Region gambar sebesar kernel
      - Kalikan elemen demi elemen dengan kernel
      - Jumlahkan semua hasil (dot product)
  5. Simpan hasil ke dalam array output
  6. Kembalikan array output dengan return
  '''

  #dapatkan dimensi
  img_height, img_width = image.shape
  kernel_height, kernel_width = kernel.shape


  # Tambahkan zero padding di citra (idk apa maksudnya but ok)
  if padding > 0:
    image = np.pad(image, padding, mode='constant', constant_values=0)

  #Hitung ukuran output
  output_height = (img_height - kernel_height) // stride + 1 #// artinya bagi dan round floor angkanya
  output_width = (img_width - kernel_width) // stride + 1

  #Buat array output di inisialisasi dengan 0 (soalnya tanpa ini ngebug dia output is not initialized)
  output = np.zeros((output_height,output_width),dtype=np.float32)

  #Loop setiap pixel di output
```
